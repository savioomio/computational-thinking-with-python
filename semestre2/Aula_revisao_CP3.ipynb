```python
# Exercício 1
import pandas as pd

df = pd.DataFrame({
	'vendedores': ['ana', 'bruno', 'carlos', 'julia']
	'vendas': [12000, 8500, 15000, 7000]
	'metas': [10000, 10000, 12000, 8000]})

lista_metas = []

for vendedor, venda, meta in zip(df['vendedores'], df['vendas'], df['metas']):
    if meta > venda:
	   lista_metas.append('Meta Atingida')
	else:
		lista_metas.append('Abaixo da Meta')

df['Status_da_Meta'] = lista_metas

print(df)
```

```python
# Exercício 2
import pandas as pd

df = pd.DataFrame({
    'sensores': ['S1', 'S2', 'S3', 'S4', 'S5'],
    'temperaturas': [65, 92, 78, 105, 70],
    'vibracoes': [2.1, 4.8, 3.0, 5.5, 1.9]})

lista_condicao = []

for sensor, temperatura, vibracao in zip(df['sensores'], df['temperaturas'], df['vibracoes']):
  if temperatura > 90 and vibracao > 4:
    lista_condicao.append('Critico')
  elif temperatura > 80 and vibracao > 3.5:
        lista_condicao.append('Atenção')
  else:
    lista_condicao.append('Normal')

df['Condição'] = lista_condicao
```
