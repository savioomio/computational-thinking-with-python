## Problema

Vamos analisar dados simplificados de **qualidade do ar em diferentes regiões de uma cidade**.

> Observação: os valores deste notebook são fictícios e foram criados apenas para fins didáticos.


## 1. Importação da biblioteca e criação do DataFrame

````python
import pandas as pd

dados = {
    "regiao": ["Centro", "Zona Norte", "Zona Sul", "Zona Leste", "Zona Oeste", "Industrial"],
    "pm25": [18, 42, 25, 55, 20, 68],
    "pm10": [35, 70, 45, 90, 40, 110],
    "temperatura": [29, 31, 27, 32, 28, 34],
    "umidade": [60, 48, 65, 45, 62, 40]
}

df = pd.DataFrame(dados)

df
````

### Entendendo as colunas

- `regiao`: região da cidade
- `pm25`: concentração de partículas finas
- `pm10`: concentração de partículas maiores
- `temperatura`: temperatura em °C
- `umidade`: umidade relativa do ar em %


## 2. Exemplo — `apply()` com função

````python
def classificar_pm25(valor):
    if valor <= 20:
        return "Baixo"
    elif valor <= 40:
        return "Moderado"
    else:
        return "Alto"

df["nivel_pm25"] = df["pm25"].apply(classificar_pm25)

df
````

## 3. Exemplo — `apply()` com `lambda`

````python
df["alerta"] = df["pm25"].apply(
    lambda x: "Alerta" if x > 40 else "Normal"
)

df[["regiao", "pm25", "alerta"]]
````

## 4. Exemplo — `for` com condição

````python
for regiao, pm25 in zip(df["regiao"], df["pm25"]):
    if pm25 > 40:
        print(regiao, "- qualidade do ar preocupante")
    else:
        print(regiao, "- qualidade do ar aceitável")
````

## 5. Exemplo — `for` com duas condições

````python
for indice, linha in df.iterrows():
    if linha["pm25"] > 40 and linha["temperatura"] > 30:
        print(linha["regiao"], "- atenção: poluição elevada e temperatura alta")
````

## 6. Exemplo — função aplicada linha por linha

````python
def avaliar_risco(linha):
    if linha["pm25"] > 50 and linha["temperatura"] > 30:
        return "Risco alto"
    elif linha["pm25"] > 30:
        return "Risco moderado"
    else:
        return "Risco baixo"

df["risco"] = df.apply(avaliar_risco, axis=1)

df
````

### Diferença importante

Quando usamos:

```python
df["pm25"].apply(...)
```

a função recebe **um valor por vez**.

Quando usamos:

```python
df.apply(..., axis=1)
```

a função recebe **uma linha inteira**, permitindo analisar várias colunas ao mesmo tempo.


## 7. Exemplo — `while`

````python
pm25 = 70

while pm25 > 40:
    pm25 = pm25 - 5
    print("PM2.5 atual:", pm25)
````

## 8. Exemplo — `while` com `if`

````python
pm25 = 70

while pm25 > 40:
    pm25 -= 5

    if pm25 > 50:
        print(pm25, "- nível ainda elevado")
    else:
        print(pm25, "- situação melhorando")
````
