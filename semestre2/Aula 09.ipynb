```python
import pandas as pd
import numpy as np

bruto = pd.DataFrame({
    "pedido":    [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008],
    "descricao": ["Notebook Dell - SP", "mouse logitech - rj", "Monitor LG - SP",
                  "  Teclado Redragon - MG ", "NOTEBOOK Lenovo - RJ", "Webcam Logitech - sp",
                  None, "Monitor Samsung - MG"],
    "preco":     ["R$ 4.200,00", "89,90", "R$ 1.350,00", "210", None,
                  "R$ 149,90", "R$ 99,00", "não informado"],
    "qtd":       ["2", "10", None, "5", "1", "4", "3", "2"],
})
bruto
```

```python
# 1. Tira as linhas em que a coluna de texto está vazia.

def remover_sem_descricao(df):
    return df.dropna(subset=["descricao"]).reset_index(drop=True)

passo = remover_sem_descricao(bruto)
```

```python
# 2. Padronizando o texto
""".strip() tira os espaços das pontas"""
""".title() coloca a primeira letra de cada palavra em maiúscula"""

def padronizar_texto(df):
    df = df.copy()
    df["descricao"] = df["descricao"].str.strip().str.title()
    return df

passo = padronizar_texto(passo)
```

```python
# Padronizando texto de uma outra forma
"""
A ideia é separar a descrição no " - ", aplicar .title() só na parte do produto e
.upper() só na parte do estado, e depois juntar de novo."""

def padronizar_texto(df):
    df = df.copy()
    partes = df["descricao"].str.strip().str.split(" - ")
    df["descricao"] = partes.str[0].str.title() + " - " + partes.str[1].str.upper()
    return df

df = padronizar_texto(df)
df
```

```python
# Criando nova coluna (Produto)
"""
.str.split() quebra o texto em uma lista de palavras
.str[0] pega a primeira.
"""

def extrair_produto(df):
    df = df.copy()
    df["produto"] = df["descricao"].str.split().str[0]
    return df

df = extrair_produto(df)
```

```python
df
```

```python
# Criando nova coluna (marca)
"""
.str.split() quebra o texto em uma lista de palavras
.str[1] pega a segunda palavra, que é a marca
"""

def extrair_marca(df):
    df = df.copy()
    df["marca"] = df["descricao"].str.split().str[1]
    return df

df = extrair_marca(df)
```

```python
# Criando nova coluna (estado)
"""
.str.split() quebra o texto em uma lista de palavras
.str[-1] pega a última palavra, que é o estado
.str.upper() deixa tudo em maiúsculas
"""

def extrair_estado(df):
    df = df.copy()
    df["estado"] = df["descricao"].str.split().str[-1].str.upper()
    return df

df = extrair_estado(df)
```

```python
# Colocar preco para float

""".str.replace("R$", "", regex=False)	Apaga o símbolo da moeda (troca "R$" por nada)	" 4.200,00"
.str.replace(".", "", regex=False)	Apaga o ponto de milhar	" 4200,00"
.str.replace(",", ".", regex=False)	Troca a vírgula decimal pelo ponto, que é o padrão do Python	" 4200.00"
.str.trip()	Tira os espaços do começo e do fim	"4200.00
to_numeric(df["preco"] - passa a coluna preco para numerico.
errors="coerce" - Vira NaN e a conversão segue
"""

def limpar_preco(df):
    df = df.copy()
    df["preco"] = (df["preco"]
                   .str.replace("R$", "", regex=False)
                   .str.replace(".", "", regex=False)   # tira o ponto de milhar
                   .str.replace(",", ".", regex=False)  # vírgula vira ponto decimal
                   .str.strip())
    df["preco"] = pd.to_numeric(df["preco"], errors="coerce")
    return df
```

```python
df = limpar_preco(df)
df
```

```python
# Tratar precos ausentes.

def tratar_preco_ausente(df):
    df = df.copy()
    mediana_produto = df.groupby("produto")["preco"].transform("median")
    df["preco"] = df["preco"].fillna(mediana_produto)
    return df
df = tratar_preco_ausente(df)
```

```python
# 8. Converter qtd para inteiro

"""
errors="coerce" - Vira NaN e a conversão segue
fillna(0) - preenche os valores nulos com 0
.astype(int) - converte para inteiro
"""

def tratar_qtd(df):
    df = df.copy()
    df["qtd"] = pd.to_numeric(df["qtd"], errors="coerce").fillna(0).astype(int)
    return df

df = tratar_qtd(df)
```

```python
# Calculando total a partir de preco e qtd

def calcular_total(df):
    df = df.copy()
    df["total"] = df.apply(lambda linha: linha["preco"] * linha["qtd"], axis=1)
    return df

df = calcular_total(df)
df
```

```python
# Criando a coluna frete

df["frete"] = df.apply(
    lambda linha: 0 if linha["total"] > 1000 or linha["estado"] == "SP" else 25,
    axis=1)
df
```

```python
# Dropando a coluna descricao
def remover_descricao(df):
    df = df.copy()
    df = df.drop(columns="descricao")
    return df

df = remover_descricao(df)
df
```
